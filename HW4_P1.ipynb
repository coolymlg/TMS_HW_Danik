```python
from collections import Counter

my_text = input("Введите желаемый текст: ")
my_text = my_text.lower()

my_words = (len(my_text.split()))
print ("Кол-во слов: ",my_words)

my_wordsun = len(list(dict.fromkeys(my_text.split())))
print ("Кол-во уникальных слов: ",my_wordsun)



my_often = Counter(my_text.split()).most_common(1)
print ("Самое частое слово и его кол-во повторений: ",my_often)

my_split = my_text.split()

lenght = 0
for i in range(len(my_split)):
  lenght += len(my_split[i])

avg_len = lenght / (len(my_text.split()))
print ("Средняя длинна слова: ",avg_len)
```

```python
import random
guess = random.randrange(1, 100)
print(guess)
user_input = int(input("Введите число для угадывания (от 1 до 100): "))
while True:
  if user_input == guess:
    print ("Верно! Ты угадал!")
    break
  if user_input > guess:
    user_input = int(input("Не верно, твоё число больше загаданного, попробуй еще раз: "))
  else:
    user_input = int(input("Не верно, твоё число меньше загаданного, попробуй еще раз: "))
```

```python
from IPython.display import clear_output
import time

seconds = int(input("Введи время в секундах: "))
while seconds > 0:
  clear_output(wait=True)
  min = seconds // 60
  sec = seconds % 60
  print(f"{min:02}:{sec:02}")
  time.sleep(1)
  seconds-= 1
print("Таймер окончен.")

```
